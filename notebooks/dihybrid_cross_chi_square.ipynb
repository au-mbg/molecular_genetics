# Interactive Dihybrid Cross: Understanding Probability Rules in Genetics

## What is this tool?

This interactive notebook helps you visualize and understand how probability rules work in genetics, specifically for a **dihybrid cross** (AaBb × AaBb). You'll explore how to calculate the probability of different genotypes and phenotypes using the **product rule** and **sum rule**.

## Learning objectives

By working through this tool, you will:

1. **Understand the product rule**: Learn how to multiply independent probabilities across different loci
2. **Understand the sum rule**: Learn how to add probabilities of mutually exclusive outcomes
3. **Connect theory to visualization**: See how Punnett squares and branch diagrams represent the same genetic cross in different ways
4. **Practice probability calculations**: Work through increasingly complex genetics problems

## How to use this notebook

The notebook is divided into 4 sections, each with interactive controls:

### Part 1: Single Genotype Probability (Product Rule)
- **What to do**: Select one genotype at locus A (AA, Aa, or aa) and one at locus B (BB, Bb, or bb)
- **What you'll learn**: How to calculate the probability of a specific genotype by multiplying probabilities at each locus
- **Key concept**: P(AaBb) = P(Aa) × P(Bb) = 1/2 × 1/2 = 1/4
- **Try this**: Select AaBb and verify that exactly 4 cells (4/16) are highlighted in yellow in the Punnett square

### Part 2: Multiple Genotypes (Sum Rule)
- **What to do**: Select multiple genotypes from the list (hold Ctrl/Cmd to select more than one)
- **What you'll learn**: How to calculate the probability of getting ANY of several genotypes by adding their individual probabilities
- **Key concept**: P(AaBb OR AABb) = P(AaBb) + P(AABb) = 4/16 + 2/16 = 6/16
- **Try this**: Select AaBb and AABb together and see how the probabilities add up

### Part 3: Phenotype Probability (Product Rule + Sum Rule)
- **What to do**: Select a phenotype combination (e.g., A_B_ means dominant for both traits)
- **What you'll learn**: How phenotypes involve BOTH rules—product rule for the phenotype probability, sum rule because multiple genotypes produce the same phenotype
- **Key concept**: P(A_B_) = P(A_) × P(B_) = 3/4 × 3/4 = 9/16, which includes 4 different genotypes (AABB, AABb, AaBB, AaBb)
- **Try this**: Select A_B_ and count how many different genotypes (yellow boxes) contribute to this phenotype

### Part 4: Complex Queries
- **What to do**: Choose from pre-programmed genetics problems
- **What you'll learn**: How to approach complex questions that require strategic use of both probability rules
- **Key concept**: Sometimes using the complement rule (1 - P(opposite)) is easier than direct calculation
- **Try this**: Compare "At least one dominant" using both the direct method and the complement method

## Understanding the visualizations

### Punnett Square (left)
- Shows all 16 possible offspring from the cross AaBb × AaBb
- Each cell = one possible genotype with probability 1/16
- Yellow cells = your selected genotype(s)
- **Count the yellow cells** to verify the probability calculation (e.g., 4 yellow cells = 4/16 = 1/4)

### Branch Diagram (right)
- Shows the same information as a decision tree
- Follow paths **left to right**: Start → Locus A outcome → Locus B outcome → Final genotype
- Each branch shows the probability at that step
- Final boxes show the **product** of probabilities along the path
- Yellow boxes (with red borders) = your selected genotype(s)
- **This is why we multiply**: Following one path through independent events requires multiplying probabilities

## Tips for learning

1. **Start simple**: Begin with Part 1, try different single genotypes, and verify the counts
2. **Compare visualizations**: For each selection, look at BOTH the Punnett square and branch diagram—they tell the same story in different ways
3. **Predict before selecting**: Before changing the dropdown, try to predict how many cells will be highlighted
4. **Test your understanding**: Can you explain WHY AaBb appears 4 times but AABb only appears 2 times?
5. **Work through Part 4**: These complex queries are similar to exam questions—make sure you understand both solution methods

## Common mistakes to avoid

- **Don't confuse genotype and phenotype**: AA, Aa, and aa are genotypes; A_ (dominant) and aa (recessive) are phenotypes
- **Don't add when you should multiply**: Use product rule for independent events (different loci), sum rule for mutually exclusive outcomes (different genotypes)
- **Don't forget 2/4 for heterozygotes**: Aa appears twice as often as AA or aa because there are two ways to get it (A from mom, a from dad OR a from mom, A from dad)

## Questions to test yourself

1. Why does the branch diagram have 9 final outcomes but the Punnett square has 16 cells?
2. What's the probability of getting at least one heterozygous locus (Aa or Bb)?
3. If you select phenotype A_bb, which genotypes contribute and why?
4. How many different genotypes exist for a dihybrid cross? How many different phenotypes?

*Answers: 1) The branch diagram shows 9 genotypic classes; the Punnett square shows all 16 individual offspring. 2) Use complement: 1 - P(AABB or aabb) = 1 - 2/16 = 14/16 = 7/8. 3) AAbb and Aabb (both have dominant A and recessive bb). 4) 9 genotypes, 4 phenotypes (with complete dominance).*


```python
#@title Interactive Dihybrid Cross: Product & Sum Rules with Punnett Squares & Branch Diagrams

print("=" * 70)
print("INTERACTIVE DIHYBRID CROSS: PRODUCT & SUM RULE CALCULATOR")
print("WITH PUNNETT SQUARES & BRANCH DIAGRAMS")
print("=" * 70)
print("\nCross: AaBb × AaBb")
print("Calculate probabilities with visual aids!\n")

# ============================================================================
# VISUALIZATION FUNCTIONS
# ============================================================================

def draw_punnett_square_fixed(highlight_cells=None):
    """
    Draw a 4x4 Punnett square for AaBb × AaBb with CORRECTED highlighting.
    """
    fig, ax = plt.subplots(figsize=(11, 11))
    ax.set_xlim(-1, 5)
    ax.set_ylim(-1, 5)
    ax.axis('off')

    gametes = ['AB', 'Ab', 'aB', 'ab']

    # Create genotype grid
    genotype_grid = []
    for row_gamete in gametes:
        row = []
        for col_gamete in gametes:
            allele_A1 = row_gamete[0]
            allele_A2 = col_gamete[0]
            if allele_A1 == 'a' and allele_A2 == 'A':
                geno_A = 'Aa'
            elif allele_A1 == 'A' and allele_A2 == 'a':
                geno_A = 'Aa'
            elif allele_A1 == 'A' and allele_A2 == 'A':
                geno_A = 'AA'
            else:
                geno_A = 'aa'

            allele_B1 = row_gamete[1]
            allele_B2 = col_gamete[1]
            if allele_B1 == 'b' and allele_B2 == 'B':
                geno_B = 'Bb'
            elif allele_B1 == 'B' and allele_B2 == 'b':
                geno_B = 'Bb'
            elif allele_B1 == 'B' and allele_B2 == 'B':
                geno_B = 'BB'
            else:
                geno_B = 'bb'

            genotype = geno_A + geno_B
            row.append(genotype)
        genotype_grid.append(row)

    # Draw borders
    ax.plot([0, 4], [0, 0], 'k-', linewidth=3)
    ax.plot([0, 4], [4, 4], 'k-', linewidth=3)
    ax.plot([0, 0], [0, 4], 'k-', linewidth=3)
    ax.plot([4, 4], [0, 4], 'k-', linewidth=3)

    for i in range(1, 4):
        ax.plot([i, i], [0, 4], 'k-', linewidth=1)
        ax.plot([0, 4], [i, i], 'k-', linewidth=1)

    # Label gametes
    for i, g in enumerate(gametes):
        ax.text(i + 0.5, 4.3, g, ha='center', va='center',
                fontsize=16, weight='bold', color='blue',
                bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.7))

    for i, g in enumerate(gametes):
        ax.text(-0.3, 3.5 - i, g, ha='center', va='center',
                fontsize=16, weight='bold', color='blue',
                bbox=dict(boxstyle='round', facecolor='lightblue', alpha=0.7))

    highlighted_count = 0

    # Fill genotypes
    for row in range(4):
        for col in range(4):
            genotype = genotype_grid[row][col]
            should_highlight = False
            if highlight_cells:
                should_highlight = genotype in highlight_cells

            x = col
            y = 3 - row

            if should_highlight:
                color = 'yellow'
                highlighted_count += 1
            else:
                color = 'white'

            rect = mpatches.Rectangle((x, y), 1, 1,
                                      linewidth=0, facecolor=color, alpha=0.7)
            ax.add_patch(rect)

            ax.text(x + 0.5, y + 0.5, genotype,
                   ha='center', va='center', fontsize=13, weight='bold')

    title = 'Punnett Square: AaBb × AaBb'
    if highlight_cells:
        title += f'\n{highlighted_count} cells highlighted (yellow)'
    ax.set_title(title, fontsize=18, weight='bold', pad=30)

    info_text = f"Total cells: 16\nEach cell = 1/16"
    if highlight_cells:
        info_text += f"\n\nSelected:\n" + "\n".join(highlight_cells[:4])
        if len(highlight_cells) > 4:
            info_text += f"\n... +{len(highlight_cells)-4} more"
        info_text += f"\n\nProbability:\n{highlighted_count}/16"

    ax.text(4.5, 2, info_text, fontsize=11, va='center',
            bbox=dict(boxstyle='round', facecolor='lightyellow', alpha=0.8))

    plt.tight_layout()  # FIXED: was plt.tight_tight()
    return fig, highlighted_count

def draw_branch_diagram(highlight_genotypes=None):
    """
    Draw a forked-line/branch diagram for AaBb × AaBb with accurate probabilities.
    FIXED: Proper spacing for all 9 distinct genotype paths.
    """
    fig, ax = plt.subplots(figsize=(16, 12))
    ax.set_xlim(0, 13)
    ax.set_ylim(0, 11)
    ax.axis('off')

    # Starting point
    ax.text(0.8, 5.5, 'AaBb × AaBb', fontsize=13, weight='bold', ha='center',
            bbox=dict(boxstyle='round', facecolor='lightblue', edgecolor='black', linewidth=2))

    # Locus A: Aa × Aa outcomes
    genotypes_A = ['AA', 'Aa', 'aa']
    probs_A = [1/4, 2/4, 1/4]
    y_positions_A = [8.5, 5.5, 2.5]

    # Draw locus A branches
    for geno_A, prob_A, y_A in zip(genotypes_A, probs_A, y_positions_A):
        ax.plot([1.5, 3], [5.5, y_A], 'k-', linewidth=2)
        ax.text(3.5, y_A, f'{geno_A}\nP = {prob_A.as_integer_ratio()[0]}/{prob_A.as_integer_ratio()[1]}',
                fontsize=11, weight='bold', ha='center', va='center',
                bbox=dict(boxstyle='round', facecolor='lightyellow', edgecolor='black'))

    # Locus B: Bb × Bb outcomes (from each A outcome)
    genotypes_B = ['BB', 'Bb', 'bb']
    probs_B = [1/4, 2/4, 1/4]

    # For each locus A outcome, draw locus B branches with proper spacing
    for i, (geno_A, prob_A, y_A) in enumerate(zip(genotypes_A, probs_A, y_positions_A)):
        y_offsets = [0.8, 0, -0.8]

        for j, (geno_B, prob_B, y_offset) in enumerate(zip(genotypes_B, probs_B, y_offsets)):
            y_B = y_A + y_offset

            # Draw branch from A to B
            ax.plot([4.5, 6.5], [y_A, y_B], 'gray', linewidth=1.5, alpha=0.7)

            # Combined genotype and probability
            combined_geno = geno_A + geno_B
            combined_prob = prob_A * prob_B
            prob_fraction = f"{int(combined_prob * 16)}/16"

            # Check if this genotype should be highlighted
            is_highlighted = (highlight_genotypes and combined_geno in highlight_genotypes)
            bgcolor = 'yellow' if is_highlighted else 'lightgreen'
            edgecolor = 'red' if is_highlighted else 'black'
            linewidth_box = 2.5 if is_highlighted else 1

            # Draw locus B outcome box
            ax.text(7.5, y_B, f'{geno_B}\nP = {probs_B[j].as_integer_ratio()[0]}/{probs_B[j].as_integer_ratio()[1]}',
                   fontsize=10, ha='center', va='center',
                   bbox=dict(boxstyle='round', facecolor='lightcyan', edgecolor='black'))

            # Draw branch from B to final genotype
            ax.plot([8.5, 9.5], [y_B, y_B], 'gray', linewidth=1.5, alpha=0.7)

            # Draw final genotype box with product calculation
            prob_calc = f"P = {prob_A.as_integer_ratio()[0]}/{prob_A.as_integer_ratio()[1]} × {probs_B[j].as_integer_ratio()[0]}/{probs_B[j].as_integer_ratio()[1]} = {prob_fraction}"
            ax.text(11, y_B, f'{combined_geno}\n{prob_calc}',
                   fontsize=10, ha='center', va='center', weight='bold' if is_highlighted else 'normal',
                   bbox=dict(boxstyle='round', facecolor=bgcolor, edgecolor=edgecolor,
                            linewidth=linewidth_box, alpha=0.8))

    # Add section labels
    ax.text(3.5, 10, 'Locus A\n(Aa × Aa)', fontsize=12, ha='center', weight='bold',
            bbox=dict(boxstyle='round', facecolor='lavender', edgecolor='blue', linewidth=2))
    ax.text(7.5, 10, 'Locus B\n(Bb × Bb)', fontsize=12, ha='center', weight='bold',
            bbox=dict(boxstyle='round', facecolor='lavender', edgecolor='blue', linewidth=2))
    ax.text(11, 10, 'Final Genotype\n(Product Rule)', fontsize=12, ha='center', weight='bold',
            bbox=dict(boxstyle='round', facecolor='lightcoral', edgecolor='red', linewidth=2))

    ax.set_title('Branch Diagram (Forked-Line Method): AaBb × AaBb\nFollow branches left → right, multiply probabilities',
                fontsize=16, weight='bold', pad=20)
    plt.tight_layout()
    return fig

# ============================================================================
# PROBABILITY DEFINITIONS
# ============================================================================

locus_A_probs = {'AA': 1/4, 'Aa': 2/4, 'aa': 1/4}
locus_B_probs = {'BB': 1/4, 'Bb': 2/4, 'bb': 1/4}

# ============================================================================
# PART 1: SINGLE GENOTYPE PROBABILITY (Product Rule)
# ============================================================================
print("### PART 1: SINGLE GENOTYPE PROBABILITY (Product Rule) ###\n")

def calculate_single_genotype(geno_A, geno_B):
    """
    Calculate probability of a specific genotype using product rule.
    """
    p_A = locus_A_probs[geno_A]
    p_B = locus_B_probs[geno_B]
    p_combined = p_A * p_B

    genotype = geno_A + geno_B

    print(f"Selected genotype: {genotype}")
    print(f"\n--- Using PRODUCT RULE ---")
    print(f"P({geno_A}) at locus A = {p_A} = {p_A.as_integer_ratio()[0]}/{p_A.as_integer_ratio()[1]}")
    print(f"P({geno_B}) at locus B = {p_B} = {p_B.as_integer_ratio()[0]}/{p_B.as_integer_ratio()[1]}")
    print(f"\nP({genotype}) = P({geno_A}) × P({geno_B})")
    print(f"           = {p_A} × {p_B}")
    print(f"           = {p_combined}")

    numerator = p_combined.as_integer_ratio()[0]
    denominator = p_combined.as_integer_ratio()[1]
    out_of_16 = int(p_combined * 16)
    print(f"           = {numerator}/{denominator}")
    print(f"           = {out_of_16}/16")
    print(f"\nThis genotype appears in {out_of_16} out of 16 cells.\n")

    # Draw visualizations
    fig_punnett, actual_count = draw_punnett_square_fixed(highlight_cells=[genotype])
    if actual_count == out_of_16:
        print(f"✓ Punnett Square: {actual_count} cells highlighted (matches {out_of_16}/16)")
    plt.show()

    fig_branch = draw_branch_diagram(highlight_genotypes=[genotype])
    print(f"\n✓ Branch Diagram: Yellow box shows {genotype} with P = {out_of_16}/16")
    plt.show()

interact(calculate_single_genotype,
         geno_A=Dropdown(options=['AA', 'Aa', 'aa'], value='Aa', description='Locus A:'),
         geno_B=Dropdown(options=['BB', 'Bb', 'bb'], value='Bb', description='Locus B:'))

# ============================================================================
# PART 2: MULTIPLE GENOTYPES (Sum Rule)
# ============================================================================
print("\n### PART 2: MULTIPLE GENOTYPES (Sum Rule) ###\n")

all_genotypes = []
for gA in ['AA', 'Aa', 'aa']:
    for gB in ['BB', 'Bb', 'bb']:
        all_genotypes.append(f"{gA}{gB}")

def calculate_multiple_genotypes(selected):
    """
    Calculate probability of getting ANY of the selected genotypes using sum rule.
    """
    if len(selected) == 0:
        print("Please select at least one genotype!")
        return

    print(f"Selected genotypes: {', '.join(selected)}\n")
    print("--- Using PRODUCT RULE for each, then SUM RULE ---\n")

    total_prob = 0
    calculations = []

    for genotype in selected:
        geno_A = genotype[:2]
        geno_B = genotype[2:]

        p_A = locus_A_probs[geno_A]
        p_B = locus_B_probs[geno_B]
        p_combined = p_A * p_B

        total_prob += p_combined
        out_of_16 = int(p_combined * 16)

        calculations.append(f"P({genotype}) = {p_A} × {p_B} = {p_combined} = {out_of_16}/16")

    for calc in calculations:
        print(calc)

    print(f"\n--- Applying SUM RULE ---")
    print(f"P(any of selected) = " + " + ".join([f"P({g})" for g in selected]))

    total_out_of_16 = int(total_prob * 16)
    print(f"                   = {total_out_of_16}/16")
    print(f"                   = {total_prob}")
    print(f"                   ≈ {total_prob:.4f} or {total_prob*100:.2f}%")
    print(f"\nExpect {total_out_of_16} highlighted cells in visualizations.\n")

    # Draw visualizations
    fig_punnett, actual_count = draw_punnett_square_fixed(highlight_cells=list(selected))
    if actual_count == total_out_of_16:
        print(f"✓ Punnett Square: {actual_count} cells highlighted (matches {total_out_of_16}/16)")
    plt.show()

    fig_branch = draw_branch_diagram(highlight_genotypes=list(selected))
    print(f"\n✓ Branch Diagram: Yellow boxes show all selected genotypes")
    plt.show()

interact(calculate_multiple_genotypes,
         selected=SelectMultiple(
             options=all_genotypes,
             value=['AaBb', 'AABb'],
             description='Genotypes:',
             rows=9
         ))

# ============================================================================
# PART 3: PHENOTYPE PROBABILITY (Product Rule + Sum Rule)
# ============================================================================
print("\n### PART 3: PHENOTYPE PROBABILITY (Product Rule + Sum Rule) ###\n")

pheno_A_probs = {'A_': 3/4, 'aa': 1/4}
pheno_B_probs = {'B_': 3/4, 'bb': 1/4}

def calculate_phenotype(pheno_A, pheno_B):
    """
    Calculate probability of a phenotype combination.
    """
    p_A = pheno_A_probs[pheno_A]
    p_B = pheno_B_probs[pheno_B]
    p_combined = p_A * p_B

    print(f"Selected phenotype: {pheno_A} {pheno_B}")
    print(f"\n--- Using PRODUCT RULE ---")
    print(f"P({pheno_A}) = {p_A} = {p_A.as_integer_ratio()[0]}/{p_A.as_integer_ratio()[1]}")
    print(f"P({pheno_B}) = {p_B} = {p_B.as_integer_ratio()[0]}/{p_B.as_integer_ratio()[1]}")
    print(f"\nP({pheno_A} {pheno_B}) = P({pheno_A}) × P({pheno_B})")
    print(f"              = {p_A} × {p_B}")
    print(f"              = {p_combined}")

    out_of_16 = int(p_combined * 16)
    print(f"              = {out_of_16}/16")

    print(f"\nGenotypes that give this phenotype:")
    if pheno_A == 'A_':
        genos_A = ['AA', 'Aa']
    else:
        genos_A = ['aa']

    if pheno_B == 'B_':
        genos_B = ['BB', 'Bb']
    else:
        genos_B = ['bb']

    contributing_genotypes = []
    for gA in genos_A:
        for gB in genos_B:
            genotype = gA + gB
            contributing_genotypes.append(genotype)
            print(f"  - {genotype}")

    print(f"\n--- Using SUM RULE ---")
    print(f"Sum of all contributing genotypes = {out_of_16}/16")
    print(f"\nExpect {out_of_16} highlighted cells.\n")

    # Draw visualizations
    fig_punnett, actual_count = draw_punnett_square_fixed(highlight_cells=contributing_genotypes)
    if actual_count == out_of_16:
        print(f"✓ Punnett Square: {actual_count} cells highlighted (matches {out_of_16}/16)")
    plt.show()

    fig_branch = draw_branch_diagram(highlight_genotypes=contributing_genotypes)
    print(f"\n✓ Branch Diagram: Yellow boxes show all genotypes for {pheno_A} {pheno_B}")
    plt.show()

interact(calculate_phenotype,
         pheno_A=Dropdown(options=['A_', 'aa'], value='A_', description='Pheno A:'),
         pheno_B=Dropdown(options=['B_', 'bb'], value='B_', description='Pheno B:'))

# ============================================================================
# PART 4: COMPLEX QUERIES (Combined Rules)
# ============================================================================
print("\n### PART 4: COMPLEX QUERIES (Product + Sum Rules) ###\n")

def complex_query(query_type):
    """
    Pre-defined complex queries that require both product and sum rules.
    """
    contributing_genotypes = []

    if query_type == "At least one dominant (A_ or B_)":
        print("Question: What is the probability of showing at least one dominant trait?\n")
        print("--- Solution ---")
        print("This includes phenotypes: A_B_, A_bb, aaB_")
        print("(Everything EXCEPT aabb)\n")
        print("Method 1 - Direct sum:")
        print("P(A_B_) = 3/4 × 3/4 = 9/16")
        print("P(A_bb) = 3/4 × 1/4 = 3/16")
        print("P(aaB_) = 1/4 × 3/4 = 3/16")
        print("Sum = 9/16 + 3/16 + 3/16 = 15/16\n")
        print("Method 2 - Complement:")
        print("P(at least one dominant) = 1 - P(aabb)")
        print("                         = 1 - (1/4 × 1/4)")
        print("                         = 1 - 1/16")
        print("                         = 15/16")
        print(f"\nAnswer: 15/16 = {15/16:.4f} = 93.75%\n")

        for gA in ['AA', 'Aa', 'aa']:
            for gB in ['BB', 'Bb', 'bb']:
                genotype = gA + gB
                if genotype != 'aabb':
                    contributing_genotypes.append(genotype)

    elif query_type == "Exactly one dominant trait":
        print("Question: What is the probability of exactly one dominant trait?\n")
        print("--- Solution ---")
        print("This includes: A_bb OR aaB_ (but NOT A_B_ or aabb)\n")
        print("P(A_bb) = P(A_) × P(bb) = 3/4 × 1/4 = 3/16")
        print("P(aaB_) = P(aa) × P(B_) = 1/4 × 3/4 = 3/16")
        print("\nUsing SUM RULE (mutually exclusive):")
        print("P(exactly one dominant) = P(A_bb) + P(aaB_)")
        print("                        = 3/16 + 3/16")
        print("                        = 6/16 = 3/8")
        print(f"\nAnswer: 3/8 = {3/8:.4f} = 37.5%\n")

        for gA in ['AA', 'Aa']:
            contributing_genotypes.append(gA + 'bb')
        for gB in ['BB', 'Bb']:
            contributing_genotypes.append('aa' + gB)

    elif query_type == "Both recessive (aabb)":
        print("Question: What is the probability of both traits showing recessive?\n")
        print("--- Solution ---")
        print("This is phenotype aabb only.\n")
        print("Using PRODUCT RULE:")
        print("P(aabb) = P(aa) × P(bb)")
        print("        = 1/4 × 1/4")
        print("        = 1/16")
        print(f"\nAnswer: 1/16 = {1/16:.4f} = 6.25%\n")

        contributing_genotypes = ['aabb']

    elif query_type == "Heterozygous at both loci (AaBb)":
        print("Question: What is the probability of genotype AaBb?\n")
        print("--- Solution ---")
        print("Using PRODUCT RULE:")
        print("P(AaBb) = P(Aa) × P(Bb)")
        print("        = 1/2 × 1/2")
        print("        = 1/4 = 4/16")
        print(f"\nAnswer: 1/4 = {1/4:.4f} = 25%\n")

        contributing_genotypes = ['AaBb']

    elif query_type == "Homozygous at least one locus":
        print("Question: Probability of being homozygous at one or both loci?\n")
        print("--- Solution ---")
        print("Easier to use complement: 1 - P(heterozygous at both)\n")
        print("P(Aa) = 1/2, P(Bb) = 1/2")
        print("P(AaBb) = 1/2 × 1/2 = 1/4 = 4/16")
        print("\nP(homozygous at ≥1 locus) = 1 - P(AaBb)")
        print("                           = 1 - 4/16")
        print("                           = 12/16 = 3/4")
        print(f"\nAnswer: 3/4 = {3/4:.4f} = 75%\n")

        for gA in ['AA', 'Aa', 'aa']:
            for gB in ['BB', 'Bb', 'bb']:
                genotype = gA + gB
                if genotype != 'AaBb':
                    contributing_genotypes.append(genotype)

    # Draw visualizations
    if contributing_genotypes:
        fig_punnett, actual_count = draw_punnett_square_fixed(highlight_cells=contributing_genotypes)
        print(f"✓ Punnett Square: {actual_count}/16 cells highlighted")
        plt.show()

        fig_branch = draw_branch_diagram(highlight_genotypes=contributing_genotypes)
        print(f"\n✓ Branch Diagram: Yellow boxes show all genotypes for this query")
        plt.show()

interact(complex_query,
         query_type=Dropdown(
             options=[
                 "At least one dominant (A_ or B_)",
                 "Exactly one dominant trait",
                 "Both recessive (aabb)",
                 "Heterozygous at both loci (AaBb)",
                 "Homozygous at least one locus"
             ],
             value="At least one dominant (A_ or B_)",
             description='Query:',
             style={'description_width': '150px'},
             layout=widgets.Layout(width='500px')
         ))

print("\n" + "=" * 70)
print("SUMMARY:")
print("• PRODUCT RULE: Multiply probabilities across independent loci")
print("• SUM RULE: Add probabilities of mutually exclusive outcomes")
print("• Punnett Square: Count yellow cells to verify")
print("• Branch Diagram: Follow path, multiply probabilities, add probabilities of mutually exclusive outcomes")
print("=" * 70)
```

# Interactive Chi-Square Test: Statistical Analysis of Genetic Ratios

## What is this tool?

This interactive notebook helps you understand how to use the **chi-square (χ²) goodness-of-fit test** to determine whether observed genetic data match expected Mendelian ratios.

## Why do we need statistical tests in genetics?

When you perform a genetic cross (e.g., Aa × Aa), you **expect** a 3:1 ratio of dominant to recessive phenotypes. But real experimental data rarely show EXACTLY 75 dominant and 25 recessive in 100 offspring. Random sampling causes variation.

The chi-square test answers: **"Is the difference between my observed data and the expected ratio due to random chance, or is something else going on (like the trait isn't actually Mendelian)?"**

## Learning objectives

By working through this tool, you will:

1. **Master probability rules**: Practice applying product and sum rules to genetics problems
2. **Understand the chi-square test**: Learn what the test measures and how to interpret p-values
3. **Develop statistical intuition**: See how small deviations are expected, but large deviations suggest problems
4. **Apply to real genetics**: Practice with monohybrid (3:1), dihybrid (9:3:3:1), and custom ratios

## How to use this notebook

The notebook is divided into 3 sections:

### Part 1: Chi-Square Test - Monohybrid Cross (Interactive)
- **What to do**: Use the sliders to adjust observed counts of dominant and recessive phenotypes
- **What you'll learn**: How the chi-square test compares observed data to the expected 3:1 Mendelian ratio
- **Key concept**: χ² = Σ[(Observed - Expected)²/Expected]; if χ² is large, the data don't fit the model
- **Try this**:
  - Set dominant = 150, recessive = 50 (perfect 3:1) → χ² should be very small, p > 0.05, ACCEPT null hypothesis
  - Set dominant = 180, recessive = 20 (deviation from 3:1) → χ² should be large, p < 0.05, REJECT null hypothesis
- **Watch the p-value**: If p > 0.05, the data fit the expected ratio (random variation). If p < 0.05, the data significantly deviate (something's wrong)

### Part 2: Chi-Square Test - Dihybrid Cross (Interactive)
- **What to do**: Adjust four sliders for the 9:3:3:1 phenotypic ratio in a dihybrid cross
- **What you'll learn**: How chi-square works with more complex ratios (more degrees of freedom)
- **Key concept**: With 4 categories, df = 3, critical value = 7.815 (at α = 0.05)
- **Try this**:
  - Set A_B_ = 225, A_bb = 75, aaB_ = 75, aabb = 25 (perfect 9:3:3:1) → should accept
  - Change A_B_ to 180 and aabb to 60 (shift toward recessive) → should reject
- **Notice**: The bar chart makes deviations easy to see visually

### Part 3: Custom Ratio Test (Interactive)
- **What to do**: Set both observed counts AND expected ratios for up to 4 classes
- **What you'll learn**: How to test ANY genetic ratio (epistasis, incomplete dominance, linkage, etc.)
- **Examples to try**:
  - **1:1 ratio** (testcross): Set ratio = 1:1:0:0, adjust counts accordingly
  - **1:2:1 ratio** (incomplete dominance): Set ratio = 1:2:1:0
  - **12:3:1 ratio** (epistasis): Set ratio = 12:3:1:0
- **Powerful tool**: This lets you test hypotheses about non-standard inheritance patterns

## Interpreting chi-square results

### The chi-square statistic (χ²)
- **χ² = 0**: Observed data EXACTLY match expected (never happens in real experiments)
- **χ² is small** (< critical value): Differences are likely due to random chance → **ACCEPT null hypothesis**
- **χ² is large** (≥ critical value): Differences are too large to be random → **REJECT null hypothesis**

### The p-value
- **p > 0.05**: The probability of seeing this much deviation by chance alone is > 5% → data fit the model → **ACCEPT**
- **p ≤ 0.05**: The probability of seeing this much deviation by chance is ≤ 5% → data don't fit the model → **REJECT**
- **Think of p-value as**: "If the model is correct, how likely is it I'd see data this different from expected?"

### Degrees of freedom (df)
- **df = number of categories - 1**
- Monohybrid (2 categories): df = 1, critical value = 3.841
- Dihybrid (4 categories): df = 3, critical value = 7.815
- More categories = higher critical value needed to reject

## Tips for learning

1. **Start with perfect ratios**: Set observed counts to EXACTLY match expected ratios and verify χ² ≈ 0 and p ≈ 1.0
2. **Introduce small deviations**: Adjust counts slightly (e.g., 155:45 instead of 150:50) and see χ² increase, but p might still be > 0.05
3. **Create large deviations**: Make big changes (e.g., 180:20) and watch p-value drop below 0.05
4. **Compare visual and statistical**: Look at the bar charts—can you predict whether the test will accept or reject before seeing the numbers?
5. **Test real hypotheses**: Use Part 5 to test different genetic models (epistasis, linkage) with custom ratios

## Common mistakes to avoid

- **Don't confuse χ² and p-value**: χ² is the test statistic; p-value is the probability. Focus on the p-value for decision-making
- **Don't forget sample size matters**: Small samples can show large deviations by chance; large samples make the test more powerful
- **Don't misinterpret "accept"**: We never "prove" a model is correct; we only say the data are consistent with it
- **Don't misinterpret "reject"**: Rejection could mean wrong model, but also could mean experimental error, contamination, etc.

## Advanced: When chi-square might mislead you

- **Expected counts < 5**: Chi-square is unreliable when any expected category has fewer than 5 individuals
- **Non-random sampling**: If your sample is biased (e.g., you selectively removed sick plants), chi-square can't detect this
- **Multiple testing**: Testing many hypotheses increases false positives—use Bonferroni correction if needed

```python
#@title Interactive Cross: Chi-square test


# ============================================================================
# PART 1: INTERACTIVE CHI-SQUARE TEST (MONOHYBRID)
# ============================================================================
print("\n### PART 1: CHI-SQUARE TEST - MONOHYBRID CROSS (Interactive) ###\n")

def chi_square_monohybrid(obs_dominant, obs_recessive):
    """
    Chi-square test for monohybrid cross (3:1 ratio).
    """
    # Expected 3:1 ratio
    total = obs_dominant + obs_recessive
    exp_dominant = total * 3/4
    exp_recessive = total * 1/4

    observed = np.array([obs_dominant, obs_recessive])
    expected = np.array([exp_dominant, exp_recessive])

    # Chi-square calculation
    chi2_stat = np.sum((observed - expected)**2 / expected)
    df = 1
    p_value = 1 - stats.chi2.cdf(chi2_stat, df)
    critical_value = stats.chi2.ppf(0.95, df)

    # Display results
    print(f"Total offspring: {total}")
    print(f"Expected (3:1): Dominant = {exp_dominant:.2f}, Recessive = {exp_recessive:.2f}")
    print(f"Observed: Dominant = {obs_dominant}, Recessive = {obs_recessive}\n")

    print(f"χ² statistic: {chi2_stat:.4f}")
    print(f"Degrees of freedom: {df}")
    print(f"p-value: {p_value:.4f}")
    print(f"Critical value (α=0.05): {critical_value:.4f}\n")

    if p_value > 0.05:
        print("✓ ACCEPT null hypothesis (p > 0.05)")
        print("  → Data FIT the expected 3:1 Mendelian ratio")
    else:
        print("✗ REJECT null hypothesis (p ≤ 0.05)")
        print("  → Data DO NOT fit the expected 3:1 ratio")

    # Visualization
    fig, ax = plt.subplots(figsize=(8, 5))
    x = np.arange(2)
    width = 0.35
    ax.bar(x - width/2, expected, width, label='Expected', alpha=0.8, color='skyblue')
    ax.bar(x + width/2, observed, width, label='Observed', alpha=0.8, color='coral')
    ax.set_xlabel('Phenotype')
    ax.set_ylabel('Count')
    ax.set_title(f'Monohybrid Cross: χ² = {chi2_stat:.3f}, p = {p_value:.4f}')
    ax.set_xticks(x)
    ax.set_xticklabels(['Dominant', 'Recessive'])
    ax.legend()
    plt.tight_layout()
    plt.show()

interact(chi_square_monohybrid,
         obs_dominant=IntSlider(value=150, min=0, max=300, step=1, description='Dominant:'),
         obs_recessive=IntSlider(value=50, min=0, max=100, step=1, description='Recessive:'))

# ============================================================================
# PART 2: INTERACTIVE CHI-SQUARE TEST (DIHYBRID)
# ============================================================================
print("\n### PART 2: CHI-SQUARE TEST - DIHYBRID CROSS (Interactive) ###\n")

def chi_square_dihybrid(obs_AB, obs_Ab, obs_aB, obs_ab):
    """
    Chi-square test for dihybrid cross (9:3:3:1 ratio).
    """
    # Expected 9:3:3:1 ratio
    total = obs_AB + obs_Ab + obs_aB + obs_ab
    ratio = np.array([9, 3, 3, 1])
    expected = (ratio / ratio.sum()) * total

    observed = np.array([obs_AB, obs_Ab, obs_aB, obs_ab])

    # Chi-square calculation
    chi2_stat = np.sum((observed - expected)**2 / expected)
    df = 3
    p_value = 1 - stats.chi2.cdf(chi2_stat, df)
    critical_value = stats.chi2.ppf(0.95, df)

    # Display results
    print(f"Total offspring: {total}")
    print(f"Expected (9:3:3:1): {expected}")
    print(f"Observed: [A_B_={obs_AB}, A_bb={obs_Ab}, aaB_={obs_aB}, aabb={obs_ab}]\n")

    print(f"χ² statistic: {chi2_stat:.4f}")
    print(f"Degrees of freedom: {df}")
    print(f"p-value: {p_value:.4f}")
    print(f"Critical value (α=0.05): {critical_value:.4f}\n")

    if p_value > 0.05:
        print("✓ ACCEPT null hypothesis (p > 0.05)")
        print("  → Data FIT the expected 9:3:3:1 ratio")
    else:
        print("✗ REJECT null hypothesis (p ≤ 0.05)")
        print("  → Data DO NOT fit the 9:3:3:1 ratio")

    # Visualization
    fig, ax = plt.subplots(figsize=(10, 5))
    x = np.arange(4)
    width = 0.35
    ax.bar(x - width/2, expected, width, label='Expected', alpha=0.8, color='skyblue')
    ax.bar(x + width/2, observed, width, label='Observed', alpha=0.8, color='coral')
    ax.set_xlabel('Phenotype Class')
    ax.set_ylabel('Count')
    ax.set_title(f'Dihybrid Cross (9:3:3:1): χ² = {chi2_stat:.3f}, p = {p_value:.4f}')
    ax.set_xticks(x)
    ax.set_xticklabels(['A_B_', 'A_bb', 'aaB_', 'aabb'])
    ax.legend()
    plt.tight_layout()
    plt.show()

interact(chi_square_dihybrid,
         obs_AB=IntSlider(value=225, min=0, max=400, step=5, description='A_B_:'),
         obs_Ab=IntSlider(value=75, min=0, max=200, step=5, description='A_bb:'),
         obs_aB=IntSlider(value=75, min=0, max=200, step=5, description='aaB_:'),
         obs_ab=IntSlider(value=25, min=0, max=100, step=5, description='aabb:'))

# ============================================================================
# PART 3: CUSTOM RATIO CHI-SQUARE TEST
# ============================================================================
print("\n### PART 3: CUSTOM RATIO CHI-SQUARE TEST (Interactive) ###\n")
print("Test any custom expected ratio (e.g., 1:1, 1:2:1, 12:3:1, etc.)\n")

def custom_chi_square(class1, class2, class3, class4, ratio1, ratio2, ratio3, ratio4):
    """
    Chi-square test for any custom expected ratio.
    """
    observed = np.array([class1, class2, class3, class4])
    ratio = np.array([ratio1, ratio2, ratio3, ratio4])

    # Remove zero-ratio classes
    mask = ratio > 0
    observed = observed[mask]
    ratio = ratio[mask]

    total = observed.sum()
    expected = (ratio / ratio.sum()) * total

    # Chi-square calculation
    chi2_stat = np.sum((observed - expected)**2 / expected)
    df = len(observed) - 1
    p_value = 1 - stats.chi2.cdf(chi2_stat, df)
    critical_value = stats.chi2.ppf(0.95, df)

    print(f"Total offspring: {total}")
    print(f"Expected ratio: {ratio}")
    print(f"Expected counts: {expected}")
    print(f"Observed counts: {observed}\n")

    print(f"χ² statistic: {chi2_stat:.4f}")
    print(f"Degrees of freedom: {df}")
    print(f"p-value: {p_value:.4f}")
    print(f"Critical value (α=0.05): {critical_value:.4f}\n")

    if p_value > 0.05:
        print("✓ ACCEPT null hypothesis (p > 0.05)")
        print(f"  → Data FIT the expected {':'.join(map(str, ratio))} ratio")
    else:
        print("✗ REJECT null hypothesis (p ≤ 0.05)")
        print(f"  → Data DO NOT fit the expected {':'.join(map(str, ratio))} ratio")

    # Visualization
    fig, ax = plt.subplots(figsize=(10, 5))
    x = np.arange(len(observed))
    width = 0.35
    ax.bar(x - width/2, expected, width, label='Expected', alpha=0.8, color='skyblue')
    ax.bar(x + width/2, observed, width, label='Observed', alpha=0.8, color='coral')
    ax.set_xlabel('Class')
    ax.set_ylabel('Count')
    ax.set_title(f'Custom Ratio: χ² = {chi2_stat:.3f}, p = {p_value:.4f}')
    ax.set_xticks(x)
    ax.set_xticklabels([f'Class {i+1}' for i in range(len(observed))])
    ax.legend()
    plt.tight_layout()
    plt.show()

interact(custom_chi_square,
         class1=IntSlider(value=120, min=0, max=400, step=5, description='Class 1:'),
         class2=IntSlider(value=40, min=0, max=400, step=5, description='Class 2:'),
         class3=IntSlider(value=40, min=0, max=400, step=5, description='Class 3:'),
         class4=IntSlider(value=0, min=0, max=400, step=5, description='Class 4:'),
         ratio1=IntSlider(value=9, min=0, max=20, step=1, description='Ratio 1:'),
         ratio2=IntSlider(value=3, min=0, max=20, step=1, description='Ratio 2:'),
         ratio3=IntSlider(value=4, min=0, max=20, step=1, description='Ratio 3:'),
         ratio4=IntSlider(value=0, min=0, max=20, step=1, description='Ratio 4:'))

print("\n" + "=" * 70)
print("TIP: Adjust sliders to test your understanding!")
print("Try deviating from expected ratios to see when χ² becomes significant.")
print("=" * 70)
```
